```python
# Student grade A - above 90 marks, B - above 80 marks, C - above 70 marks, D - above 60 marks, E - above 50 marks, F - below 50 marks
# if elif else format
marks = 88
if marks > 90:
    grade = "A"
elif marks > 80:
    grade = "B"
elif marks > 70:
    grade = "C"
elif marks > 60:
    grade = "D"
elif marks > 50:
    grade = "E"
else:
    grade = "F" 
print("The grade for marks", marks, "is:", grade)   
```
